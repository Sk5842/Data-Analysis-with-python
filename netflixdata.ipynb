  ***Categoriing Vote_Aerage column***
* we will replace Vote_Average and make four category like Popular, average ,below average, not popular to describe it more using catigorize_col() funcion provide above

```python
df.describe()
```

```python
def catigorize_col(df,col,labels):
    edges=[df[col].describe()['min'],
           df[col].describe()['25%'],
           df[col].describe()['50%'],
           df[col].describe()['75%'],
           df[col].describe()['max']]
    df[col]=pd.cut(df[col],edges,labels=labels,duplicates='drop')
    return df
```

```python
labels=['not_popular','below_avg','average','popular']

catigorize_col(df,'Vote_Average', labels)

df['Vote_Average'].unique()
```
